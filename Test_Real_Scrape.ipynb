```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://www.worldometers.info/world-population/population-by-country/'
```

```python
response = requests.get(url)
html = response.text
```

```python
soup = BeautifulSoup(html, 'html.parser')
```

```python
"""
We have a 'table-structure' => <thead> + <tbody>
    <thead>: contains the column names
    <tbody>: contains the row information

To work with a smaller chunck of sourcecode that only contains the row data:
    we take only the chunck we need: body = soup.find('tbody')            
"""
```

```python
columns = soup.find_all('span')
columns
```

```python
column_names = [column.text.strip() for column in columns]
column_names
```

```python
body = soup.find('tbody')
```

```python
"""
We have <tr> for a full row
    + <td> for a cell in a row
"""
```

```python
# Row Container
rows = body.find_all('tr')
```

```python
scraped_data = []
```

```python
# First we test the first 3 rows if it is as expected !!!
for row in rows[:3]:
    print(row.text.strip())
```

```python

for row in rows:
    
    cells = row.find_all('td') 
    # For the second loop we cannot just use 'row' => it is 1 html element so NOT ITERABLE !!!
    # So we define change it into a container of elements => EACH ELEMENT IS ITERABLE !!!
    row_list = [cell.text.strip() for cell in cells]
    scraped_data.append(row_list)
```

```python
import pandas as pd
```

```python
world_data = pd.DataFrame(data = scraped_data, columns = column_names)
```
